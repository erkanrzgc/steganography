```python
import hashlib
import json
import os
import re
import sys
from pathlib import Path

from google.colab import files

SOURCE_SHA256 = '8677a46f3fb7840b01b46b2b1355a286c9d30296ccf271519fe54bc57d8e300a'
if any(n == "core" or n.startswith("core.") for n in sys.modules):
    raise RuntimeError("Use a fresh kernel; preserve every consumed attempt")
uploaded = files.upload()  # Select the accompanying sources.json only.
if set(uploaded) != {'png_bmp_colab_native_training_sources_20261107.json'}:
    raise ValueError("Select sources.json only")
raw = uploaded['png_bmp_colab_native_training_sources_20261107.json']
if len(raw) > 4 * 1024**2 or hashlib.sha256(raw).hexdigest() != SOURCE_SHA256:
    raise ValueError("Reviewed source packet hash/extent mismatch")
packet = json.loads(raw)
if (set(packet) != {"schema", "base_commit", "files"}
        or packet["schema"] != "colab-native-source-package-v1"
        or packet["base_commit"] != '2f41bfe2e95febcaab80f80ab4875c72073c3a32'
        or not 3 <= len(packet["files"]) <= 513 or "LICENSE" not in packet["files"]):
    raise ValueError("Reviewed source packet schema mismatch")
prepared = {}
for name, row in packet["files"].items():
    allowed = re.fullmatch(
        r"(?:core|steganography)/(?:[a-zA-Z0-9_]+/){0,3}[a-zA-Z0-9_]+\.py", name
    )
    if name != "LICENSE" and allowed is None:
        raise ValueError("Reviewed source path mismatch")
    if set(row) != {"utf8", "sha256"} or not isinstance(row["utf8"], str):
        raise ValueError("Reviewed source entry mismatch")
    content = row["utf8"].encode()
    if len(content) > 1024**2 or hashlib.sha256(content).hexdigest() != row["sha256"]:
        raise ValueError("Reviewed source file hash/extent mismatch")
    prepared[name] = content
source_root = Path("/content/spatial-colab-native-training-source-20261107")
if any(p.is_symlink() for p in (source_root, *source_root.parents)):
    raise ValueError("Source directory symlink")
source_root.mkdir(mode=0o700)
for name, content in prepared.items():
    target = source_root / name
    target.parent.mkdir(mode=0o700, parents=True, exist_ok=True)
    fd = os.open(target, os.O_WRONLY | os.O_CREAT | os.O_EXCL | os.O_NOFOLLOW, 0o600)
    with os.fdopen(fd, "wb") as stream:
        stream.write(content)
sys.path.insert(0, str(source_root))
print("Reviewed source package installed:", SOURCE_SHA256)
```

```python
VERIFY_APPROVED_PERMISSION = False
if VERIFY_APPROVED_PERMISSION:
    selected = files.upload()  # Only the later, explicitly approved profile-permission.json.
    if set(selected) != {"profile-permission.json"}:
        raise ValueError("Select the approved permission file only")
    permission_path = source_root / "profile-permission.json"
    artifacts.put(permission_path, selected["profile-permission.json"])
    job = artifacts.document(protocol, 4 * artifacts.MIB)
    controller.permission(job, EXPECTED_CONTRACT, permission_path)
    profile_permission_verified = True
```

```python
RUN_APPROVED_PROFILE = False
CACHE_PACKAGE_SHA256 = ""  # Fill only from the separately reviewed train-only packet.
if RUN_APPROVED_PROFILE:
    if not profile_permission_verified:
        raise ValueError("Bound permission must be verified before any media upload")
    artifacts.identity(CACHE_PACKAGE_SHA256)
    selected = files.upload()  # Select train-cache.json; no reserved/development pixels.
    if set(selected) != {"train-cache.json"}:
        raise ValueError("Select train-cache.json only")
    cache_packet = source_root / "train-cache.json"
    artifacts.put(cache_packet, selected["train-cache.json"], handoff.CACHE_LIMIT)
    split = profile.engine.split_document(PAYLOAD["split"])
    cache = handoff.install_cache(cache_packet, CACHE_PACKAGE_SHA256, split,
                                  source_root / "cache")
    output = source_root / "profile-output"
    result = controller.run(source_root, protocol, output,
                            permission_path=permission_path, cache=cache.root)
    controller.audit(source_root, protocol, output)
    print(json.dumps(result, sort_keys=True, indent=2))
```

```python
DOWNLOAD_VERIFIED_EVIDENCE = False
if DOWNLOAD_VERIFIED_EVIDENCE:
    checked = controller.audit(source_root, protocol, output)
    if checked["status"] != "completed":
        raise ValueError("Unavailable evidence cannot support fitting approval")
    archive = output / "verified-profile.zip"
    checksum = checked["output_files"][archive.name]["sha256"]
    handoff.verify_evidence(archive, checksum, EXPECTED_CONTRACT)
    files.download(str(archive))
    files.download(str(output / "closed.json"))
```

```python
PREPARE_FIT_PROPOSAL = False
if PREPARE_FIT_PROPOSAL:
    from core import spatial_colab_fit as training

    controller.audit(source_root, protocol, output)
    selected = files.upload()  # Measured setup/review/export debits, never assumed zero.
    if set(selected) != {"external-debit.json"}:
        raise ValueError("Select external-debit.json only")
    external_debit_path = source_root / "external-debit.json"
    artifacts.put(external_debit_path, selected["external-debit.json"])
    fit_protocol = source_root / "fit-contract.json"
    training.freeze(source_root, output, artifacts.document(external_debit_path), fit_protocol)
    fit_job, fit_binding, fit_limits = controller.source_contract(source_root, fit_protocol)
    fit_permission_verified = False
    print("Fit proposal:", fit_binding)
    print("Projected seconds:", fit_job["payload"]["projection_seconds"])
    print("Remaining seconds:", fit_limits.seconds, "— later fitting permission required")
```

```python
VERIFY_APPROVED_FIT = False
if VERIFY_APPROVED_FIT:
    selected = files.upload()  # Later explicit fitting approval; profile approval is insufficient.
    if set(selected) != {"fit-permission.json"}:
        raise ValueError("Select fit-permission.json only")
    fit_permission = source_root / "fit-permission.json"
    artifacts.put(fit_permission, selected["fit-permission.json"])
    controller.permission(fit_job, fit_binding, fit_permission)
    fit_permission_verified = True
```

```python
RUN_APPROVED_FIT = False
DEVELOPMENT_PACKAGE_SHA256 = ""
if RUN_APPROVED_FIT:
    if not fit_permission_verified:
        raise ValueError("Separate fit permission must precede development upload")
    artifacts.identity(DEVELOPMENT_PACKAGE_SHA256)
    selected = files.upload()  # Exactly twenty development crops, no reserved pixels.
    if set(selected) != {"development-cache.json"}:
        raise ValueError("Select development-cache.json only")
    development_packet = source_root / "development-cache.json"
    artifacts.put(development_packet, selected["development-cache.json"], handoff.CACHE_LIMIT)
    handoff.install_development(cache, split, development_packet, DEVELOPMENT_PACKAGE_SHA256)
    fit_output = source_root / "fit-output"
    fit_result = controller.run(source_root, fit_protocol, fit_output, cache=cache.root,
                                permission_path=fit_permission, profile_out=output)
    controller.audit(source_root, fit_protocol, fit_output)
    print("Execution:", fit_result["status"])
    if fit_result["status"] == "completed":
        for cell in fit_result["worker"]["cells"]:
            print("Seed", cell["seed"], "final train/development AUC:",
                  cell["final"]["train"]["auc"], cell["final"]["development"]["auc"])
```

```python
DOWNLOAD_VERIFIED_TRAINING = False
if DOWNLOAD_VERIFIED_TRAINING:
    checked = controller.audit(source_root, fit_protocol, fit_output)
    if checked["status"] != "completed":
        raise ValueError("Keep unavailable evidence private; do not claim a complete fit")
    archive = fit_output / "verified-training.zip"
    checksum = checked["output_files"][archive.name]["sha256"]
    handoff.verify_evidence(archive, checksum, fit_binding, max_members=4097)
    files.download(str(archive))
    for cell in checked["worker"]["cells"]:
        for checkpoint in cell["checkpoints"]:
            relative = "work/fit/" + checkpoint["name"]
            path = fit_output / relative
            expected_sha = checked["output_files"][relative]["sha256"]
            if handoff.stream_hash(path, fit_limits.file_bytes) != expected_sha:
                raise ValueError("Checkpoint changed before export")
            files.download(str(path))
    files.download(str(fit_output / "closed.json"))
```
